# Transformers

In this worksheet we will examine the transformer model as implemented by Google Gemma

First we will download the Google Gemma 2B model

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained('google/gemma-2b')
model = AutoModelForCausalLM.from_pretrained('google/gemma-2b')
```

Now we examine the model by simply printing it

```python
print(model)
```

**Q**: From the output above what are the following hyperparameters

* Hidden Layer Size
* Vocabulary Size
* Number of Layers
* Internal Size of Feed-Forward Layer

Now let's try to use the model on some text

```python
text = "The HSBC bank is by the river bank"
encoded_input = tokenizer(text, return_tensors='pt')
print(encoded_input)
output = model(**encoded_input, output_hidden_states=True)
print(output.keys())
```

```python
print(output.logits.shape)
```
